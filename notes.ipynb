```python
# get a list of participants
participants = df['Name'].unique()
participants
```

```python
df['HomeScoreDiff'] = df['Home Points']-df['Away Points']
df['Predict_Home_Win'] = df['HomeScoreDiff'].apply(lambda x: 1 if x > 0 else 0)
df
```

```python
# after running Rich's web scraping tool, import results for 2024
res = pd.read_csv('data/SixNationsResults.csv')
res.head()
```

```python
# filter to include 2024 rounds 1 and 2 only
res2024 = res[(res['Year']==2024) & (res['Round'].isin([1, 2]))]
res2024
```

```python
# flag whether home or away is winner
res2024['HomeScoreDiff'] = res2024['FinalScoreHome']-res2024['FinalScoreAway']
res2024['HomeWin'] = res2024['HomeScoreDiff'].apply(lambda x: 1 if x > 0 else 0)

# another way of calculating above
res2024['HomeWin2'] = 0
res2024.loc[(res2024['FinalScoreHome']>res2024['FinalScoreAway']), 'HomeWin2'] = 1

# add column for match number
res2024['Match'] = res2024.groupby('Round')['Round'].rank(method='first', ascending=True)
res2024['Match'] = res2024['Match'].apply(np.int64)
res2024
```

```python
# tidy up tables before merging (drop columns)
res2024 = res2024.drop(columns = ['Year', 'Date', 'Stadium', 'HTScoreHome', 'HTScoreAway', 'HomeScoreDiff', 'HomeWin2'])
res2024
```

```python
df = df.drop(columns = ['HomeScoreDiff'])
```

```python
# merge tables
df2 = pd.merge(df, res2024, how="left", on=['Round', 'Match'])
df2

# need to add a match index to join, might also need to join on home team names at later date? or this could be a seperate table with key to join to each table
```

```python
# create home and away score differences
df2['HomePredDiff'] = -(df2['Home Points'] - df2['FinalScoreHome']).abs()
df2['AwayPredDiff'] = -(df2['Away Points'] - df2['FinalScoreAway']).abs()
df2

# add a winning bonus (-10 points if result is incorrect)
df2['WinningBonus'] = -10
df2.loc[(df2['Predict_Home_Win']==df2['HomeWin']), 'WinningBonus'] = 0
df2

# correct score bonus
df2['CorrectHomeScore'] = 0
df2.loc[(df2['Home Points']==df2['FinalScoreHome']), 'CorrectHomeScore'] = 1
df2['CorrectAwayScore'] = 0
df2.loc[(df2['Away Points']==df2['FinalScoreAway']), 'CorrectAwayScore'] = 1
df2

#sense check
#df2.loc[(df2['CorrectAwayScore']==1)]
```

```python
# total difference of predicted vs actual scores
pred_score_diff = ['HomePredDiff', 'AwayPredDiff']
df2['TotalPredDiff'] = df2[pred_score_diff].sum(axis=1)
df2

# correct score bonus
correct_scores = ['CorrectHomeScore', 'CorrectAwayScore']
df2['BothScoresCorrect'] = df2[correct_scores].sum(axis=1)
df2['CorrectScoreBonus'] = 0
df2.loc[(df2['BothScoresCorrect']==2), 'CorrectAwayScore'] = 20
df2
```

```python
# closest bonus (smallest difference gets +5 points, but this will change after each game)
df2['ClosestScoreBonus'] = 0
df2.loc[df2['TotalPredDiff'].idxmax(), 'ClosestScoreBonus'] = 5
df2

df2.sort_values('ClosestScoreBonus', ascending=False)


# 3 correct results per round bonus - change this to be a seperate table instead and join on to round score table
df3 = df2.groupby(['Round', 'Name'])['WinningBonus'].sum().reset_index(name='SumWinningBonus')
df3

df3['CorrectRoundBonus'] = 0
df3.loc[(df3['SumWinningBonus']==0), 'CorrectRoundBonus'] = 5
df3
```

```python
# calculate final scores (CorrectRoundBonus not included here, should ClosestScoreBonus be removed here too??)
calculate_final_score = ['TotalPredDiff', 'WinningBonus', 'CorrectScoreBonus', 'ClosestScoreBonus']
df2['FinalScore'] = df2[calculate_final_score].sum(axis=1)
df2
```

```python
# make a simpler table, i.e. by round and by total?

# round scores
cols = df2[['Round', 'Name', 'FinalScore']]
cols

rnd_grp = cols.groupby(['Round', 'Name']).sum()
rnd_grp

round_score = pd.merge(rnd_grp, df3, how='left', on=['Round', 'Name'])
round_score = round_score[['Round', 'Name', 'FinalScore', 'CorrectRoundBonus']]

round_score['TotalRoundScore'] = round_score[['FinalScore', 'CorrectRoundBonus']].sum(axis=1)
round_score
```

```python
# some results are -5 points off in Round 1 i.e. Ellie, JC, Keith, Leigh, Stubert
round_score[(round_score['Round']==1)]
df2[(df2['Name']=='JC') & (df2['Round']==1)]

# all round 2 scores are correct
round_score[(round_score['Round']==2)]
```

```python
# overall score
names = df2[['Name', 'FinalScore']]

total_score = names.groupby(['Name']).sum().rename(columns={'FinalScore':'TotalScore'})
total_score

# a lot if people 5 points out again lol (Serith 10 points out)
```
